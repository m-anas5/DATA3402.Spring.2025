```python
import sys

#1. Create a `matrix` class with the following properties:
    #* It can be initialized in 2 ways:
        #1. with arguments `n` and `m`, the size of the matrix. A newly instanciated matrix will contain all zeros.
        #2. with a list of lists of values. Note that since we are using lists of lists to implement matrices, it is possible that not all rows have the same number of columns. Test explicitly that the matrix is properly specified.
    #* Matrix instances `M` can be indexed with `M[i][j]` and `M[i,j]`.
class Matrix:

    def __init__(self, n=None, m=None, matrix_data=None):
            if n is not None and m is not None:
                self.n = n
                self.m = m
                self.matrix = [[0] * m for _ in range(n)]
            elif matrix_data is not None:
                first_row_length = len(matrix_data[0])  
                for row in matrix_data:
                    if len(row) != first_row_length:  
                        raise ValueError("All rows must have the same number of columns.")
                    self.matrix = matrix_data
                    self.n = len(matrix_data)
                    self.m = first_row_length
            else:
                raise ValueError("Matrix must be initialized with either (n, m) or a list of lists.")

    def __repr__(self):
        return f"Matrix({self.n}, {self.m}, {self.matrix})"

    def print_matrix(self):
        for row in self.matrix:
            print(row)

    def __getitem__(self, index):
        if isinstance(index, tuple):
            i, j = index
            return self.matrix[i][j]
        elif isinstance(index, int):
            return self.matrix[index]
        else:
            raise TypeError("Invalid index type")

#2. Add the following methods:
    #* `shape()`: returns a tuple `(n,m)` of the shape of the matrix.
    #* `transpose()`: returns a new matrix instance which is the transpose of the matrix.
    #* `row(n)` and `column(n)`: that return the nth row or column of the matrix M as a new appropriately shaped matrix object.
    #* `to_list()`: which returns the matrix as a list of lists.
    #*  `block(n_0,n_1,m_0,m_1)` that returns a smaller matrix located at the n_0 to n_1 columns and m_0 to m_1 rows.

    def shape(self):
        return self.n, self.m

    def transpose(self):
        transposed_matrix = []
        for i in range(self.m):  
            new_row = []
        
            for j in range(self.n):  
                new_row.append(self.matrix[j][i])
    

            transposed_matrix.append(new_row)

        return Matrix(matrix_data = transposed_matrix)

    def row(self, n):
        return Matrix(matrix_data= [self.matrix[n]])

    def column(self, n):
        nth_column = []
        for i in range(self.n): 
            element = self.matrix[i][n]
        
            nth_column.append([element])
        return Matrix(matrix_data = nth_column)

    def to_list(self):
        return self.matrix

    def block(self, n_0,n_1,m_0,m_1):
        block_matrix = []
        for i in range(n_0, n_1):

            row = self.matrix[i]
            sub_row = row[m_0:m_1]

            block_matrix.append(sub_row)
        return Matrix(matrix_data = block_matrix)

    #4. Add the following member functions to your class. Make sure to appropriately test the dimensions of the matrices to make sure the operations are correct.
    #* M.scalarmul(c): a matrix that is scalar product cM, where every element of M is multiplied by c.
    #* M.add(N): adds two matrices M and N. Don’t forget to test that the sizes of the matrices are compatible for this and all other operations.
    #* M.sub(N): subtracts two matrices M and N.
    #* M.mat_mult(N): returns a matrix that is the matrix product of two matrices M and N.
    #* M.element_mult(N): returns a matrix that is the element-wise product of two matrices M and N.
    #* M.equals(N): returns true/false if M==N.

    def scalarmu(self, c):
        new_matrix = [[c * self.matrix[i][j] for j in range(self.m)] for i in range(self.n)]
        return Matrix(matrix_data=new_matrix)

    def add(self, N):
        new_matrix = []
        if self.shape() != N.shape():
            raise ValueError("Matrix are different sizes")

        for i in range(self.n):
            new_row = []  
            for j in range(self.m):  
                new_row.append(self.matrix[i][j] + N[i][j])
            new_matrix.append(new_row)  
        return Matrix(matrix_data=new_matrix)

    def sub(self, N):
        new_matrix = []
        if self.shape() != N.shape():
            raise ValueError("Matrix are different sizes")

        for i in range(self.n):
            new_row = []  
            for j in range(self.m):  
                new_row.append(self.matrix[i][j] - N[i][j])
            new_matrix.append(new_row)  
        return Matrix(matrix_data=new_matrix)


    def mat_mult(self, N):
        if self.m != N.n:
            raise ValueError("Matrix multiplication not possible: Number of columns in first matrix must match number of rows in second matrix.")

        result = []
        for _ in range(self.n):
            row = []
            for _ in range(N.m):
                row.append(0)
            result.append(row)

        for i in range(self.n):
            for j in range(N.m):
                for k in range(self.m):  
                    result[i][j] += self.matrix[i][k] * N[k][j]

        return Matrix(matrix_data=result)


    def element_mult(self, N):
        new_matrix = []
        if self.shape() != N.shape():
            raise ValueError("Matrix are different sizes")

        for i in range(self.n):
            new_row = []  
            for j in range(self.m):  
                new_row.append(self.matrix[i][j] * N[i][j])
            new_matrix.append(new_row)  
        return Matrix(matrix_data=new_matrix)


    def equals(self, N):
        if self.shape() != N.shape():
            return False
        
        for i in range(self.n):
            for j in range(self.m):  
                if self.matrix[i][j] != N[i][j]:
                    return False 
        return True

        

#5. Overload python operators to appropriately use your functions in 4 and allow expressions like:
    #* 2*M
    #* M*2
    #* M+N
    #* M-N
    #* M*N
    #* M==N
    #* M=N

    def __add__(self, other):
        return self.add(other)

    def __sub__(self, other):
        return self.sub(other)

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            return self.scalarmu(other)
        elif isinstance(other, Matrix):
            return self.mat_mult(other)
        else:
            raise TypeError(f"Unsupported operand type for *: 'Matrix' and {type(other)}")

    def __rmul__(self, other):
        return  self.__mul__(other)

    def __eq__(self, other):
        return self.equals(other)

    
        

        



```

```python
#3. Write functions that create special matrices (note these are standalone functions, not member functions of your `matrix` class):
    #* `constant(n,m,c)`: returns a `n` by `m` matrix filled with floats of value `c`.
    #* `zeros(n,m)` and `ones(n,m)`: return `n` by `m` matrices filled with floats of value `0` and `1`, respectively.
    #* `eye(n)`: returns the n by n identity matrix. 

def constant (n, m, c):
    matrix = []
    for _ in range(n):  
        row = [] 
        for _ in range(m):  
            row.append(c)
        matrix.append(row)
    return matrix

def zeros(n, m):
    return constant(n, m, 0)

def ones(n, m):
    return constant(n, m, 1)

def eye(n):
    matrix = zeros(n, n)
    for i in range(n):
        matrix[i][i] = 1 
    return matrix


```

Write a function make_deck that returns a list of all of the cards in a standard card deck. The return should be a list of tuples of pairs of suit and value. For example the 10 of Clubs would be ('Clubs', 10) and Queen of Hearts would be ('Hearts', 'Queen'). Recall that a deck has 52 cards, divided into 4 suits (Clubs, Diamonds, Hearts, and Spades), and that each suit has 13 cards: 2 to 10, Jack, Queen, King, and Ace. Summit your solution with Lab 5.

```python
Values = [2, 3, 4, 5, 6, 7, 8, 9, 10, "Ace", "Jack", "Queen", "King"]
Suits = ["Hearts", "Clubs", "Spades", "Diamonds"]


def make_deck():
    deck = []
    for Suit in Suits:
        for value in Values:
            deck.append((Suit, value))
    return deck

D = make_deck()
print(D)
            
```
