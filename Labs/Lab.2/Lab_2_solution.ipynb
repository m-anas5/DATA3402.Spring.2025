*Exercise 8:* Write a function is called with a board and player number, takes input from the player using python's `input`, and modifies the board using your function from exercise 7. Note that you should keep asking for input until you have gotten a valid input that results in a valid move.

````python
import string

def make_game_board_matrix(matrix):
    columns = len(matrix[0])  
    rows = len(matrix)  

    print("   ", end="")  
    for col in range(1, columns + 1):
        print(f" {col}  ", end="")
    print()

    for row in range(rows):
        print("  " + " ---" * columns)  
        print(f"{string.ascii_uppercase[row]} ", end="")  

        for col in range(columns):
            print(f"| {matrix[row][col]} ", end="")  
        print("|")  
    print("  " + " ---" * columns)  

def pos_marker(board, player, x, y):
    if x < 0 or x >= len(board) or y < 0 or y >= len(board[0]):
        print("Incorrect position! Try again.")
        return False  # Return False to indicate failure

    if board[x][y] != " ":
        print("Place already taken! Try again.")
        return False  # Return False to indicate failure

    board[x][y] = 'X' if player == 1 else 'O'
    return True  # Return True to indicate success

def get_valid_move(board, player):
    rows = len(board)
    columns = len(board[0])
    
    while True:
        move = input(f"Player {player}, enter your move (e.g., A1, B3): ").strip().upper()
        
        if len(move) < 2:
            print("Invalid input format! Try again.")
            continue

        row_letter = move[0]
        col_number = move[1:]

        if row_letter not in string.ascii_uppercase[:rows] or not col_number.isdigit():
            print("Invalid row or column! Try again.")
            continue

        row = string.ascii_uppercase.index(row_letter)
        col = int(col_number) - 1  # Convert to 0-based index

        if col < 0 or col >= columns:
            print("Column out of range! Try again.")
            continue

        # Try placing the move
        if pos_marker(board, player, row, col):
            break  # Exit loop on a valid move

# **Game Loop for Both Players**
def play_game():
    board = []
    for _ in range(3):
        row = [0] * 3
        board.append(row)
    make_game_board_matrix(board)

    player = 1  # Start with Player 1

    for _ in range(len(board)): 
        get_valid_move(board, player)
        make_game_board_matrix(board)  

        player = 2 if player == 1 else 1  

````

*Exercise 9:* Use all of the previous exercises to implement a full tic-tac-toe game, where an appropriate board is drawn, 2 players are repeatedly asked for a location coordinates of where they wish to place a mark, and the game status is checked until a player wins or a draw occurs.

````python
import string

def make_game_board_matrix(matrix):
    columns = len(matrix[0])  
    rows = len(matrix)  

    print("   ", end="")  
    for col in range(1, columns + 1):
        print(f" {col}  ", end="")
    print()

    for row in range(rows):
        print("  " + " ---" * columns)  
        print(f"{string.ascii_uppercase[row]} ", end="")  

        for col in range(columns):
            print(f"| {matrix[row][col]} ", end="")  
        print("|")  
    print("  " + " ---" * columns)  

def pos_marker(board, player, x, y):
    if x < 0 or x >= len(board) or y < 0 or y >= len(board[0]):
        print("Incorrect position! Try again.")
        return False

    if board[x][y] != " ":
        print("Place already taken! Try again.")
        return False

    board[x][y] = 'X' if player == 1 else 'O'
    return True

def check_make_board(matrix):
    size = len(matrix)

    
    for i in range(size):
        if matrix[i][0] != " ":
            if all(matrix[i][j] == matrix[i][0] for j in range(size)):
                return matrix[i][0]

    
    for i in range(size):
        if matrix[0][i] != " ":
            if all(matrix[j][i] == matrix[0][i] for j in range(size)):
                return matrix[0][i]

    
    if matrix[0][0] != " ":
        if all(matrix[i][i] == matrix[0][0] for i in range(size)):
            return matrix[0][0]

    
    if matrix[0][size - 1] != " ":
        if all(matrix[i][size - i - 1] == matrix[0][size - 1] for i in range(size)):
            return matrix[0][size - 1]

    
    for row in matrix:
        if " " in row:
            return -1 

    return 0  

def get_valid_move(board, player):
    rows = len(board)
    columns = len(board[0])
    
    while True:
        move = input(f"Player {player}, enter your move (e.g., A1, B3): ").strip().upper()
        
        if len(move) < 2:
            print("Invalid input format! Try again.")
            continue

        row_letter = move[0]
        col_number = move[1:]

        if row_letter not in string.ascii_uppercase[:rows] or not col_number.isdigit():
            print("Invalid row or column! Try again.")
            continue

        row = string.ascii_uppercase.index(row_letter)
        col = int(col_number) - 1  

        if col < 0 or col >= columns:
            print("Column out of range! Try again.")
            continue

        if pos_marker(board, player, row, col):
            break  

def nxn_matrix (n):
    matrix = []
    for i in range(n):
        matrix.append([" "] * n)
    return matrix

def play_game():
    n = int(input("Enter board size "))  
    board = nxn_matrix(n)  
    make_game_board_matrix(board)

    player = 1  

    for _ in range(n * n):  
        get_valid_move(board, player)
        make_game_board_matrix(board)  

       
        result = check_make_board(board)
        if result == "X":
            print("Player 1 (X) wins! ")
            return
        elif result == "O":
            print("Player 2 (O) wins! ")
            return
        elif result == 0:
            print("It's a draw! ")
            return
        player = 2 if player == 1 else 1  


````

````python
# Test your solution here
play_game()
````
